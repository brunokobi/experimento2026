## 2. Montar a HIN (exemplo com dados sinteticos)

```python
import torch

builder = HINBuilder()
builder.add_node_type("empresa", ["emp_1", "emp_2", "emp_3"], features=torch.rand(3, 4))
builder.add_node_type("cnae", ["cnae_A", "cnae_B"])
builder.add_edge_type(
    "empresa", "atua_em", "cnae",
    edges=[("emp_1", "cnae_A"), ("emp_2", "cnae_A"), ("emp_3", "cnae_B")],
    bidirectional=True,
)
builder.stats()
```

## 3. Inspecao visual (networkx)

```python
import networkx as nx
import matplotlib.pyplot as plt

graph = builder.to_networkx()
nx.draw(graph, with_labels=True, node_size=400, font_size=8)
plt.show()
```
